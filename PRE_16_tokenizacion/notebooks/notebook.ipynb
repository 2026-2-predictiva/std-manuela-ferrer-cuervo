```python

# ============================================================
# PRE 16 - PROCESAMIENTO DE TEXTO
# Manu | Carga de documentos originales
# ============================================================

import glob

def load_data(input_directory):

    # Manu: Inicializar la lista de documentos
    sequence = []

    # Manu: Identificar los archivos disponibles
    files = sorted(glob.glob(f"{input_directory}/*"))

    for file in files:

        # Manu: Leer el contenido de cada documento
        with open(file, "rt", encoding="utf-8") as f:
            raw_text = f.read()

        sequence.append((file, raw_text))

    return sequence


# Manu: Cargar los documentos del laboratorio
sequence = load_data(input_directory="../data")

for file, text in sequence:
    print(f"{file}  {text[:70]}")

assert len(sequence) > 0, "No se encontraron documentos."

print("PASSED - Carga de documentos")
```

```python

# ============================================================
# PRE 16 - PROCESAMIENTO DE TEXTO
# Manu | Limpieza y normalización de documentos
# ============================================================

import re

def clean_text(sequence):

    # Manu: Almacenar los documentos limpios
    cleaned_sequence = []

    for file, text in sequence:

        # Manu: Reemplazar saltos de línea por espacios
        cleaned_text = re.sub(r"\n", " ", text)

        # Manu: Eliminar espacios repetidos
        cleaned_text = re.sub(r"\s+", " ", cleaned_text)

        # Manu: Quitar espacios al inicio y al final
        cleaned_text = cleaned_text.strip()

        # Manu: Convertir el texto a minúsculas
        cleaned_text = cleaned_text.lower()

        cleaned_sequence.append((file, cleaned_text))

    return cleaned_sequence


# Manu: Aplicar limpieza sobre los archivos originales
sequence = load_data(input_directory="../data")
cleaned_sequence = clean_text(sequence)

for file, text in cleaned_sequence:
    print(f"{file}  {text[:70]}")

print("PASSED - Limpieza del texto")
```

```python

# ============================================================
# PRE 16 - PROCESAMIENTO DE TEXTO
# Manu | Tokenización de documentos
# ============================================================

import nltk
from nltk.tokenize import word_tokenize

# Manu: Verificar la disponibilidad del tokenizador
try:
    nltk.data.find("tokenizers/punkt_tab")
except LookupError:
    nltk.download("punkt_tab", raise_on_error=True)


def tokenize(sequence):

    # Manu: Almacenar las listas de tokens
    tokenized_sequence = []

    for file, text in sequence:

        # Manu: Separar el texto en palabras y signos
        tokens = word_tokenize(text)

        tokenized_sequence.append((file, tokens))

    return tokenized_sequence


# Manu: Ejecutar la secuencia de procesamiento
sequence = load_data(input_directory="../data")
cleaned_sequence = clean_text(sequence)
tokenized_sequence = tokenize(cleaned_sequence)

for file, tokens in tokenized_sequence:
    print(f"{file}  {' '.join(tokens)[:70]}")

print("PASSED - Tokenización completada")
```

```python
# Remoción de datos ruidosos (Opcion A)


def filter_tokens_a(sequence):
    """Esta solucion puede perder tokens que contienen caracteres no alfabeticos"""
    filtered_sequence = []
    for file, tokens in sequence:
        filtered_tokens = [token for token in tokens if token.isalpha()]
        filtered_sequence.append((file, filtered_tokens))
    return filtered_sequence


sequence = load_data(input_directory="../files/input")
cleaned_sequence = clean_text(sequence)
tokenized_sequence = tokenize(cleaned_sequence)
filtered_sequence = filter_tokens_a(tokenized_sequence)
for file, text in filtered_sequence:
    print(f"{file}  {' '.join(text)[:70]}")
```

```python
# Remoción de datos ruidosos (Opcion B)


def filter_tokens_b(sequence):
    """Esta solucion puede perder tokens que contienen caracteres no alfabeticos"""
    filtered_sequence = []
    for file, tokens in sequence:
        filtered_tokens = [re.sub(r"[^a-zA-Z\s]", " ", token) for token in tokens]
        filtered_tokens = [re.sub(r"\s+", " ", token) for token in filtered_tokens]
        filtered_tokens = [token.strip() for token in filtered_tokens]
        filtered_tokens = [token for token in filtered_tokens if token != ""]
        filtered_sequence.append((file, filtered_tokens))
    return filtered_sequence


sequence = load_data(input_directory="../files/input")
cleaned_sequence = clean_text(sequence)
tokenized_sequence = tokenize(cleaned_sequence)
filtered_sequence = filter_tokens_b(tokenized_sequence)
for file, text in filtered_sequence:
    print(f"{file}  {' '.join(text)[:70]}")
```

```python
# Remove the stopwords

nltk.download("stopwords")


def remove_stopwords(sequence):
    stop_words = set(nltk.corpus.stopwords.words("english"))
    filtered_sequence = []
    for file, tokens in sequence:
        filtered_tokens = [token for token in tokens if token not in stop_words]
        filtered_sequence.append((file, filtered_tokens))
    return filtered_sequence


sequence = load_data(input_directory="../files/input")
cleaned_sequence = clean_text(sequence)
tokenized_sequence = tokenize(cleaned_sequence)
filtered_sequence = filter_tokens_b(tokenized_sequence)
filtered_sequence = remove_stopwords(filtered_sequence)
for file, text in filtered_sequence:
    print(f"{file}  {' '.join(text)[:70]}")
```

```python
import os
import textwrap


def save_data(output_directory, sequence):

    if not os.path.exists(output_directory):
        os.makedirs(output_directory)

    for file, tokens in sequence:
        file = file.replace("\\", "/")
        with open(
            f"{output_directory}/{file.split('/')[-1]}",
            "wt",
            encoding="utf-8",
        ) as f:
            f.write(textwrap.fill(" ".join(tokens), width=70))


sequence = load_data(input_directory="../data")
cleaned_sequence = clean_text(sequence)
tokenized_sequence = tokenize(cleaned_sequence)
filtered_sequence = filter_tokens_b(tokenized_sequence)
filtered_sequence = remove_stopwords(filtered_sequence)
```
